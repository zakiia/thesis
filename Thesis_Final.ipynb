```python
# full_aqi_pipeline_final.py
# =====================================================
# Full AQI Pipeline — Final Version
# - Includes ALL classification and regression models requested
# - Confusion matrices for classification (PLOT=True)
# - Predicted vs Actual scatter plots for regressions, including
#   Weighted Regression and Stacking Ensemble regression visuals
# =====================================================

# Optional installs (uncomment if running in a fresh environment / Colab)
# !pip install --upgrade pip
# !pip install numpy pandas scikit-learn matplotlib seaborn xgboost catboost lightgbm tensorflow pmdarima

import warnings
warnings.filterwarnings("ignore")

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, StratifiedKFold, GridSearchCV, KFold, cross_val_score
from sklearn.preprocessing import MinMaxScaler, LabelEncoder
from sklearn.linear_model import LogisticRegression, LinearRegression
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor, StackingClassifier, StackingRegressor
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.metrics import (
    accuracy_score, f1_score, precision_score, recall_score, classification_report, confusion_matrix,
    r2_score, mean_squared_error, mean_absolute_error
)

# Optional libs
HAS_XGBOOST = True
HAS_CATBOOST = True
HAS_LIGHTGBM = True
HAS_ARIMA = True
HAS_TF = True

try:
    from xgboost import XGBClassifier, XGBRegressor
except Exception as e:
    print("⚠️ xgboost not available:", e)
    HAS_XGBOOST = False

try:
    from catboost import CatBoostClassifier, CatBoostRegressor
except Exception as e:
    print("⚠️ catboost not available:", e)
    HAS_CATBOOST = False

try:
    from lightgbm import LGBMClassifier, LGBMRegressor
except Exception as e:
    print("⚠️ lightgbm not available:", e)
    HAS_LIGHTGBM = False

try:
    import pmdarima as pm
except Exception as e:
    print("⚠️ pmdarima (ARIMA) not available:", e)
    HAS_ARIMA = False

try:
    import tensorflow as tf
    from tensorflow.keras import layers, models
except Exception as e:
    print("⚠️ TensorFlow not available (NNs skipped):", e)
    HAS_TF = False

# -------------------------
# CONFIG
# -------------------------
PLOT = True   # True = show plots (confusion matrices + scatter plots); False = console-only
FILE_PATH = "/content/dataset.csv"  # change to your CSV path
RANDOM_STATE = 42

# -------------------------
# 1) Load dataset
# -------------------------
print("\n🔁 Loading dataset...")
cols = ['date', 'aqi', 'class', 'pm2.5', 'pm10', 'so2', 'co', 'no2', 'o3']
df = pd.read_csv(FILE_PATH, on_bad_lines="skip", names=cols, header=0)

# Parse date (supports 'YYYYMMDD' format commonly used)
df['date'] = pd.to_datetime(df['date'], format='%Y%m%d', errors='coerce')
df = df.dropna(subset=['date']).reset_index(drop=True)

# translate Chinese labels if any
translation_map = {
    "优": "Excellent", "良": "Good",
    "轻度污染": "Lightly Polluted", "中度污染": "Moderately Polluted",
    "重度污染": "Heavily Polluted", "严重污染": "Severely Polluted"
}
df['class'] = df['class'].map(lambda x: translation_map.get(x, x))

df = df.dropna().reset_index(drop=True)
print("✅ Dataset loaded:", df.shape)

# -------------------------
# 2) Feature engineering
# -------------------------
print("\n🧩 Feature engineering...")
df['month'] = df['date'].dt.month
df['dayofyear'] = df['date'].dt.dayofyear
df['weekday'] = df['date'].dt.weekday

pollutants = ['pm2.5', 'pm10', 'so2', 'co', 'no2', 'o3']
for p in pollutants:
    for lag in [1, 2, 3]:
        df[f'{p}_lag{lag}'] = df[p].shift(lag)
    df[f'{p}_roll3'] = df[p].rolling(window=3, min_periods=1).mean()

df = df.dropna().reset_index(drop=True)
print("✅ After feature engineering:", df.shape)

# -------------------------
# 3) Classification
# -------------------------
print("\n=== CLASSIFICATION ===")
X_cls = df.drop(columns=['date', 'aqi', 'class'])
y_cls = df['class'].astype(str)

le = LabelEncoder()
y_cls_enc = le.fit_transform(y_cls)
print("Classes:", list(le.classes_))

# Drop rare classes (<2 samples)
class_counts = pd.Series(y_cls_enc).value_counts()
rare_classes = class_counts[class_counts < 2].index.tolist()
if len(rare_classes) > 0:
    print("⚠️ Dropping rare classes:", [le.classes_[c] for c in rare_classes])
    mask = ~pd.Series(y_cls_enc).isin(rare_classes)
    X_cls = X_cls[mask]
    y_cls_enc = y_cls_enc[mask]

# Train/test split stratified
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(
    X_cls, y_cls_enc, test_size=0.2, stratify=y_cls_enc, random_state=RANDOM_STATE
)

scaler_cls = MinMaxScaler()
X_train_c = scaler_cls.fit_transform(X_train_c)
X_test_c = scaler_cls.transform(X_test_c)

# Base classification models
models_cls = []
models_cls.append((LogisticRegression(max_iter=2000, random_state=RANDOM_STATE), "Logistic Regression"))
if HAS_LIGHTGBM:
    models_cls.append((LGBMClassifier(random_state=RANDOM_STATE), "LightGBM"))
models_cls.append((RandomForestClassifier(n_estimators=200, random_state=RANDOM_STATE), "Random Forest"))
models_cls.append((DecisionTreeClassifier(random_state=RANDOM_STATE), "Decision Tree"))
models_cls.append((SVC(kernel='rbf', probability=True, random_state=RANDOM_STATE), "SVM"))

classification_results = []

def eval_classification(model, name):
    model.fit(X_train_c, y_train_c)
    y_pred = model.predict(X_test_c)
    acc = accuracy_score(y_test_c, y_pred)
    f1 = f1_score(y_test_c, y_pred, average='weighted')
    prec = precision_score(y_test_c, y_pred, average='weighted', zero_division=0)
    rec = recall_score(y_test_c, y_pred, average='weighted', zero_division=0)
    print(f"{name} → Accuracy={acc:.4f}, Precision={prec:.4f}, Recall={rec:.4f}, F1={f1:.4f}")
    if PLOT:
        cm = confusion_matrix(y_test_c, y_pred)
        plt.figure(figsize=(6,5))
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
        plt.title(f"{name} Confusion Matrix")
        plt.xlabel('Predicted'); plt.ylabel('True')
        plt.show()
    return acc, f1, model

print("\n➡️ Running base classification models...")
fitted_class_models = {}
for model, name in models_cls:
    try:
        acc, f1, fitted = eval_classification(model, name)
        classification_results.append([name, acc, f1])
        fitted_class_models[name] = fitted
    except Exception as e:
        print("⚠️", name, "failed:", e)

# Weighted Classifier (RF + Logistic)
print("\n➡️ Building Weighted Classifier (RF + Logistic)")
rf = RandomForestClassifier(n_estimators=200, random_state=RANDOM_STATE)
logr = LogisticRegression(max_iter=2000, random_state=RANDOM_STATE)
rf.fit(X_train_c, y_train_c)
logr.fit(X_train_c, y_train_c)
pred_rf = rf.predict_proba(X_test_c)
pred_logr = logr.predict_proba(X_test_c)
final_pred = np.argmax((pred_rf + pred_logr) / 2, axis=1)
acc = accuracy_score(y_test_c, final_pred)
f1 = f1_score(y_test_c, final_pred, average="weighted")
classification_results.append(["Weighted Classifier", acc, f1])
fitted_class_models["Weighted Classifier"] = ("ensemble", (rf, logr))
print(f"Weighted Classifier → Accuracy={acc:.4f}, F1={f1:.4f}")
if PLOT:
    cm = confusion_matrix(y_test_c, final_pred)
    plt.figure(figsize=(6,5))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
    plt.title("Weighted Classifier Confusion Matrix")
    plt.xlabel('Predicted'); plt.ylabel('True')
    plt.show()

# Tuned XGBoost and CatBoost + Stacking
print("\n➡️ Running tuned XGBoost / CatBoost and Stacking (if available)...")
cv_cls = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
estimators_stack = []

if HAS_XGBOOST:
    xgb_cls = XGBClassifier(use_label_encoder=False, eval_metric='mlogloss', random_state=RANDOM_STATE)
    xgb_params = {'n_estimators': [100, 200], 'max_depth': [3, 5], 'learning_rate': [0.05, 0.1]}
    xgb_gs = GridSearchCV(xgb_cls, xgb_params, cv=cv_cls, scoring='f1_weighted', n_jobs=-1)
    xgb_gs.fit(X_train_c, y_train_c)
    xgb_best = xgb_gs.best_estimator_
    print("XGBoost tuned best params:", xgb_gs.best_params_)
    estimators_stack.append(('xgb', xgb_best))
    acc, f1, fitted = eval_classification(xgb_best, "XGBoost (Tuned)")
    classification_results.append(["XGBoost (Tuned)", acc, f1])
    fitted_class_models["XGBoost (Tuned)"] = fitted

if HAS_CATBOOST:
    cat_cls = CatBoostClassifier(verbose=False, random_state=RANDOM_STATE)
    cat_params = {'iterations': [100, 200], 'learning_rate': [0.05, 0.1], 'depth': [4, 6]}
    cat_gs = GridSearchCV(cat_cls, cat_params, cv=cv_cls, scoring='f1_weighted', n_jobs=-1)
    cat_gs.fit(X_train_c, y_train_c)
    cat_best = cat_gs.best_estimator_
    print("CatBoost tuned best params:", cat_gs.best_params_)
    estimators_stack.append(('cat', cat_best))
    acc, f1, fitted = eval_classification(cat_best, "CatBoost (Tuned)")
    classification_results.append(["CatBoost (Tuned)", acc, f1])
    fitted_class_models["CatBoost (Tuned)"] = fitted

# Always include logistic as baseline in stacking
estimators_stack.append(('log', logr))

if len(estimators_stack) >= 2:
    stack_cls = StackingClassifier(estimators=estimators_stack, final_estimator=LogisticRegression(), cv=cv_cls, n_jobs=-1)
    stack_cls.fit(X_train_c, y_train_c)
    acc, f1, fitted = eval_classification(stack_cls, "Stacking Ensemble")
    classification_results.append(["Stacking Ensemble", acc, f1])
    fitted_class_models["Stacking Ensemble"] = fitted
else:
    print("Not enough estimators available for stacking.")

classification_df = pd.DataFrame(classification_results, columns=["Model", "Accuracy", "F1"]).sort_values('F1', ascending=False)
classification_df.to_csv("classification_summary.csv", index=False)
print("\n📊 Classification Summary:\n", classification_df)

# -------------------------
# 4) Regression
# -------------------------
print("\n=== REGRESSION ===")
# use date as index sorted for time ordering
df = df.set_index('date').sort_index()
X_reg = df.drop(columns=['class', 'aqi'])
y_reg = df['aqi']

# Train-test split without shuffle (time series)
X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(X_reg, y_reg, test_size=0.2, shuffle=False, random_state=RANDOM_STATE)

scaler_r = MinMaxScaler()
X_train_r = scaler_r.fit_transform(X_train_r)
X_test_r = scaler_r.transform(X_test_r)

def eval_regression_model(model, name):
    model.fit(X_train_r, y_train_r)
    y_pred = model.predict(X_test_r)
    r2 = r2_score(y_test_r, y_pred)
    rmse = mean_squared_error(y_test_r, y_pred) ** 0.5
    mae = mean_absolute_error(y_test_r, y_pred)
    print(f"{name} → R²={r2:.4f}, RMSE={rmse:.4f}, MAE={mae:.4f}")
    if PLOT:
        plt.figure(figsize=(6,6))
        sns.scatterplot(x=y_test_r, y=y_pred, alpha=0.7)
        plt.plot([y_test_r.min(), y_test_r.max()], [y_test_r.min(), y_test_r.max()], 'r--')
        plt.title(f"{name}: Predicted vs Actual")
        plt.xlabel("Actual AQI"); plt.ylabel("Predicted AQI")
        plt.show()
    return y_pred, r2


for model, name in base_regressors:
    try:
        y_pred, r2 = eval_regression_model(model, name)
        results_reg[name] = (y_pred, r2)
    except Exception as e:
        print("⚠️ Regression model", name, "failed:", e)

# BP Neural Network (Dense)
if HAS_TF:
    try:
        print("\n➡️ Training BP Neural Network (Dense)...")
        tf.keras.backend.clear_session()
        nn = models.Sequential([
            layers.Dense(64, activation='relu', input_shape=(X_train_r.shape[1],)),
            layers.Dense(32, activation='relu'),
            layers.Dense(1)
        ])
        nn.compile(optimizer='adam', loss='mse')
        nn.fit(X_train_r, y_train_r, epochs=30, batch_size=16, verbose=0)
        y_pred_nn = nn.predict(X_test_r).flatten()
        r2_nn = r2_score(y_test_r, y_pred_nn)
        results_reg['BP Neural Network'] = (y_pred_nn, r2_nn)
        print(f"BP Neural Network → R²={r2_nn:.4f}")
        if PLOT:
            plt.figure(figsize=(6,6))
            sns.scatterplot(x=y_test_r, y=y_pred_nn, alpha=0.7)
            plt.plot([y_test_r.min(), y_test_r.max()], [y_test_r.min(), y_test_r.max()], 'r--')
            plt.title("BP Neural Network: Predicted vs Actual")
            plt.xlabel("Actual AQI"); plt.ylabel("Predicted AQI")
            plt.show()
    except Exception as e:
        print("⚠️ BP NN failed:", e)
else:
    print("⚠️ TensorFlow not available: BP NN skipped")

# LSTM
if HAS_TF:
    try:
        print("\n➡️ Training LSTM...")
        X_train_seq = np.expand_dims(X_train_r, axis=1)
        X_test_seq = np.expand_dims(X_test_r, axis=1)
        tf.keras.backend.clear_session()
        lstm = models.Sequential([
            layers.LSTM(64, input_shape=(1, X_train_r.shape[1])),
            layers.Dense(1)
        ])
        lstm.compile(optimizer='adam', loss='mse')
        lstm.fit(X_train_seq, y_train_r, epochs=20, batch_size=16, verbose=0)
        y_pred_lstm = lstm.predict(X_test_seq).flatten()
        r2_lstm = r2_score(y_test_r, y_pred_lstm)
        results_reg['LSTM'] = (y_pred_lstm, r2_lstm)
        print(f"LSTM → R²={r2_lstm:.4f}")
        if PLOT:
            plt.figure(figsize=(6,6))
            sns.scatterplot(x=y_test_r, y=y_pred_lstm, alpha=0.7)
            plt.plot([y_test_r.min(), y_test_r.max()], [y_test_r.min(), y_test_r.max()], 'r--')
            plt.title("LSTM: Predicted vs Actual")
            plt.xlabel("Actual AQI"); plt.ylabel("Predicted AQI")
            plt.show()
    except Exception as e:
        print("⚠️ LSTM failed:", e)

# ARIMA
if HAS_ARIMA:
    try:
        print("\n➡️ Running ARIMA (auto_arima)...")
        split_idx = int(0.8 * len(y_reg))
        train_series = y_reg.iloc[:split_idx]
        test_series = y_reg.iloc[split_idx:]
        arima_model = pm.auto_arima(train_series, seasonal=False, stepwise=True, suppress_warnings=True)
        y_pred_arima = arima_model.predict(n_periods=len(test_series))
        r2_arima = r2_score(test_series, y_pred_arima)
        # align lengths: our regression y_test_r was created with train_test_split on X_reg earlier.
        # If lengths match, use it; otherwise, only store if equal.
        # Here we attempt to use the existing y_test_r slice to compute metrics if sizes match.
        if len(y_pred_arima) == len(y_test_r):
            results_reg['ARIMA'] = (y_pred_arima, r2_arima)
            print(f"ARIMA → R²={r2_arima:.4f}")
            if PLOT:
                plt.figure(figsize=(6,6))
                sns.scatterplot(x=y_test_r, y=y_pred_arima, alpha=0.7)
                plt.plot([y_test_r.min(), y_test_r.max()], [y_test_r.min(), y_test_r.max()], 'r--')
                plt.title("ARIMA: Predicted vs Actual")
                plt.xlabel("Actual AQI"); plt.ylabel("Predicted AQI")
                plt.show()
        else:
            # if lengths mismatch, skip adding to ensemble but print summary
            print("⚠️ ARIMA predictions length mismatch with y_test_r; skipping ARIMA in ensemble.")
    except Exception as e:
        print("⚠️ ARIMA failed:", e)
else:
    print("⚠️ pmdarima not available; ARIMA skipped")

# Weighted Regression Ensemble (simple average of available predictions)
print("\n➡️ Building Weighted Regression Ensemble (simple average of available model predictions)...")
# Collect predictions that match y_test_r length
preds = []
pred_names = []
for name, (ypred, r2) in results_reg.items():
    try:
        if len(ypred) == len(y_test_r):
            preds.append(np.array(ypred))
            pred_names.append(name)
    except Exception:
        pass

if len(preds) > 0:
    y_pred_weighted = np.mean(preds, axis=0)
    r2_weighted = r2_score(y_test_r, y_pred_weighted)
    rmse_weighted = mean_squared_error(y_test_r, y_pred_weighted) ** 0.5
    mae_weighted = mean_absolute_error(y_test_r, y_pred_weighted)
    results_reg['Weighted Regression Model'] = (y_pred_weighted, r2_weighted)
    print(f"Weighted Regression → R²={r2_weighted:.4f}, RMSE={rmse_weighted:.4f}, MAE={mae_weighted:.4f}")
    if PLOT:
        plt.figure(figsize=(6,6))
        sns.scatterplot(x=y_test_r, y=y_pred_weighted, alpha=0.7)
        plt.plot([y_test_r.min(), y_test_r.max()], [y_test_r.min(), y_test_r.max()], 'r--')
        plt.title("Weighted Regression Model: Predicted vs Actual")
        plt.xlabel("Actual AQI"); plt.ylabel("Predicted AQI")
        plt.show()
else:
    print("⚠️ No valid predictions available to compute weighted ensemble.")

# Stacking Regressor (XGBoost + CatBoost base, final estimator LinearRegression)
print("\n➡️ Fitting Stacking Regressor (if possible)...")
stack_reg = None
stack_preds = None
try:
    stack_estimators = []
    if HAS_XGBOOST:
        stack_estimators.append(('xgb', XGBRegressor(random_state=RANDOM_STATE)))
    if HAS_CATBOOST:
        stack_estimators.append(('cat', CatBoostRegressor(verbose=False, random_state=RANDOM_STATE)))
    if len(stack_estimators) >= 1:
        stack_reg = StackingRegressor(estimators=stack_estimators, final_estimator=LinearRegression(), n_jobs=-1)
        stack_reg.fit(X_train_r, y_train_r)
        y_pred_stack = stack_reg.predict(X_test_r)
        r2_stack = r2_score(y_test_r, y_pred_stack)
        results_reg['Stacking Ensemble'] = (y_pred_stack, r2_stack)
        print(f"Stacking Ensemble → R²={r2_stack:.4f}")
        if PLOT:
            plt.figure(figsize=(6,6))
            sns.scatterplot(x=y_test_r, y=y_pred_stack, alpha=0.7)
            plt.plot([y_test_r.min(), y_test_r.max()], [y_test_r.min(), y_test_r.max()], 'r--')
            plt.title("Stacking Ensemble Regression: Predicted vs Actual")
            plt.xlabel("Actual AQI"); plt.ylabel("Predicted AQI")
            plt.show()
    else:
        print("Not enough regressors available to build stacking regressor (need at least one of XGBoost/CatBoost).")
except Exception as e:
    print("⚠️ Stacking regressor failed:", e)

# Final regression summary
df_reg_results = pd.DataFrame([(m, float(r[1]) if r[1] is not None else np.nan) for m, r in results_reg.items()], columns=["Model", "R²"]).sort_values("R²", ascending=False)
df_reg_results.to_csv("regression_summary.csv", index=False)
print("\n📊 Regression Summary:\n", df_reg_results)

print("\n✅ Pipeline finished: classification_summary.csv and regression_summary.csv saved (if write permitted).")
```
